```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Configuration -------------------------------------------------------
import os, json, random
from pathlib import Path
import numpy as np
import nibabel as nib
import torch
import torch.nn as nn
import torch.nn.functional as F
from tqdm.auto import tqdm

SEED = 36; random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
IMG_SIZE = 512; CT_MIN_HU = -40.0; CT_MAX_HU = 120.0; THRESHOLD = 0.5
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device :', DEVICE)

MODEL_NAMES = ['socle', 'variante']

if Path('/kaggle/working').is_dir(): BASE = Path('/kaggle/working')
else: BASE = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
ROOT = utils.find_bhsd_root()
SPLITS = utils.load_splits(root=ROOT)
OUT_BASE = BASE / 'predictions'; OUT_BASE.mkdir(parents=True, exist_ok=True)
print('Test scanners:', len(SPLITS['test']))
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- U-Net 2D ------------------------------------------------------------
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.net = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True), nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
    def forward(self, x): return self.net(x)

class UNet2D(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, features=(32, 64, 128, 256)):
        super().__init__()
        self.downs = nn.ModuleList(); self.ups = nn.ModuleList(); self.pool = nn.MaxPool2d(2)
        ch = in_ch
        for f in features: self.downs.append(DoubleConv(ch, f)); ch = f
        self.bottleneck = DoubleConv(ch, ch * 2); ch = ch * 2
        for f in reversed(features):
            self.ups.append(nn.ConvTranspose2d(ch, f, 2, stride=2))
            self.ups.append(DoubleConv(f * 2, f)); ch = f
        self.final = nn.Conv2d(features[0], out_ch, 1)
    def forward(self, x):
        skip = []
        for down in self.downs: x = down(x); skip.append(x); x = self.pool(x)
        x = self.bottleneck(x); skip = skip[::-1]
        for i in range(0, len(self.ups), 2):
            x = self.ups[i](x); s = skip[i // 2]
            if x.shape[-2:] != s.shape[-2:]: x = F.interpolate(x, size=s.shape[-2:], mode='bilinear', align_corners=False)
            x = torch.cat([x, s], dim=1); x = self.ups[i + 1](x)
        return self.final(x)
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Charger les modeles --------------------------------------------------
MODEL_DIRS = [
    Path('/kaggle/input/m2-modeles-equipe-36'),
    BASE / 'modeles',
]

def trouver_modele(nom):
    for d in MODEL_DIRS:
        p = d / f'model_{nom}.pt'
        if p.is_file():
            return p
    raise FileNotFoundError(f'model_{nom}.pt introuvable dans {MODEL_DIRS}')

def charger_modele(nom):
    chemin = trouver_modele(nom)
    print(f'Chargement {nom} :', chemin)
    m = UNet2D().to(DEVICE)
    m.load_state_dict(torch.load(chemin, map_location=DEVICE))
    m.eval()
    print(f'Modele {nom} charge.')
    return m
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Prediction sur les 96 scanners de test (socle + variante) -----------
for nom_modele in MODEL_NAMES:
    out_dir = OUT_BASE / nom_modele
    out_dir.mkdir(parents=True, exist_ok=True)
    modele = charger_modele(nom_modele)
    with torch.no_grad():
        for cid in tqdm(SPLITS['test'], desc=f'Prediction {nom_modele}'):
            img = nib.load(ROOT / 'images' / f'{cid}.nii.gz')
            image = np.asarray(img.dataobj, dtype=np.float32)
            n_z = image.shape[2]
            pred_volume = np.zeros((n_z, image.shape[0], image.shape[1]), dtype=np.float32)
            for z in range(n_z):
                slice_img = image[:, :, z]
                slice_img = np.clip(slice_img, CT_MIN_HU, CT_MAX_HU)
                slice_img = (slice_img - 40.0) / 80.0  # (valeur - 40) / 80, comme dans l'article BHSD
                x = torch.from_numpy(slice_img[None, None]).float().to(DEVICE)
                prob = torch.sigmoid(modele(x)).cpu().numpy()[0, 0]
                pred_volume[z] = prob
            pred_mask = (pred_volume > THRESHOLD).astype(np.int16)
            # pred_volume est empile en (Z, X, Y) : on revient a l'ordre du volume source (X, Y, Z)
            pred_mask = np.transpose(pred_mask, (1, 2, 0))
            nib.save(nib.Nifti1Image(pred_mask, img.affine), str(out_dir / f'{cid}.nii.gz'))
    print(f'{nom_modele} : {len(list(out_dir.glob("*.nii.gz")))} masques -> {out_dir}')
    del modele
    torch.cuda.empty_cache()

print('\nTermine. Predictions dans :', OUT_BASE)
```
