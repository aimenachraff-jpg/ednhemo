# 03 Entrainement socle - U-Net 2D



```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Configuration -------------------------------------------------------
import os, json, random
from pathlib import Path
import numpy as np
import nibabel as nib
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from tqdm.auto import tqdm

SEED = 36
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
BATCH_SIZE = 8; EPOCHS = 50; LR = 1e-4; IMG_SIZE = 512
CT_MIN_HU = -40.0; CT_MAX_HU = 120.0; THRESHOLD = 0.5
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device :', DEVICE)

if Path('/kaggle/working').is_dir(): BASE = Path('/kaggle/working')
else: BASE = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
ROOT = utils.find_bhsd_root()
SPLITS = utils.load_splits(root=ROOT)
OUT = BASE / 'modeles'; OUT.mkdir(parents=True, exist_ok=True)
print('Train:', len(SPLITS['train']), 'Val:', len(SPLITS['val']), 'Test:', len(SPLITS['test']))
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Dataset PyTorch -----------------------------------------------------
class BHSDSliceDataset(Dataset):
    def __init__(self, root, case_ids, split, augment=False, seed=36):
        self.root = Path(root); self.case_ids = case_ids; self.split = split
        self.augment = augment; self.rng = random.Random(seed); self.slices = []
        for cid in case_ids:
            gt = nib.load(self.root / 'ground truths' / f'{cid}.nii.gz')
            mask = np.asarray(gt.dataobj).astype(np.int16)
            for z in range(mask.shape[2]):
                self.slices.append((cid, z))
    def __len__(self): return len(self.slices)
    def __getitem__(self, idx):
        cid, z = self.slices[idx]
        img = nib.load(self.root / 'images' / f'{cid}.nii.gz')
        gt = nib.load(self.root / 'ground truths' / f'{cid}.nii.gz')
        image = np.asarray(img.dataobj, dtype=np.float32)
        mask = np.asarray(gt.dataobj).astype(np.int16)
        img_slice = np.clip(image[:, :, z], CT_MIN_HU, CT_MAX_HU)
        img_slice = (img_slice - 40.0) / 80.0  # (valeur - 40) / 80, comme dans l'article BHSD
        mask_slice = mask[:, :, z] > 0
        if self.augment and self.rng.random() < 0.5:
            img_slice = np.fliplr(img_slice); mask_slice = np.fliplr(mask_slice)
        if self.augment:
            from scipy import ndimage
            angle = self.rng.uniform(-10, 10)
            img_slice = ndimage.rotate(img_slice, angle, reshape=False, order=1, mode='nearest')
            mask_slice = ndimage.rotate(mask_slice.astype(np.float32), angle, reshape=False, order=0, mode='nearest') > 0.5
        return (torch.from_numpy(img_slice[None]).float(), torch.from_numpy(mask_slice[None]).float())
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- U-Net 2D ------------------------------------------------------------
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.net = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True), nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
    def forward(self, x): return self.net(x)

class UNet2D(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, features=(32, 64, 128, 256)):
        super().__init__()
        self.downs = nn.ModuleList(); self.ups = nn.ModuleList(); self.pool = nn.MaxPool2d(2)
        ch = in_ch
        for f in features: self.downs.append(DoubleConv(ch, f)); ch = f
        self.bottleneck = DoubleConv(ch, ch * 2); ch = ch * 2
        for f in reversed(features):
            self.ups.append(nn.ConvTranspose2d(ch, f, 2, stride=2))
            self.ups.append(DoubleConv(f * 2, f)); ch = f
        self.final = nn.Conv2d(features[0], out_ch, 1)
    def forward(self, x):
        skip = []
        for down in self.downs: x = down(x); skip.append(x); x = self.pool(x)
        x = self.bottleneck(x); skip = skip[::-1]
        for i in range(0, len(self.ups), 2):
            x = self.ups[i](x); s = skip[i // 2]
            if x.shape[-2:] != s.shape[-2:]: x = F.interpolate(x, size=s.shape[-2:], mode='bilinear', align_corners=False)
            x = torch.cat([x, s], dim=1); x = self.ups[i + 1](x)
        return self.final(x)
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Perte Dice + BCE ----------------------------------------------------
class DiceBCELoss(nn.Module):
    def __init__(self, smooth=1e-6):
        super().__init__(); self.smooth = smooth; self.bce = nn.BCEWithLogitsLoss()
    def forward(self, logits, targets):
        bce = self.bce(logits, targets)
        probs = torch.sigmoid(logits)
        inter = (probs * targets).sum()
        dice = 1 - (2. * inter + self.smooth) / (probs.sum() + targets.sum() + self.smooth)
        return bce + dice
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- DataLoaders ---------------------------------------------------------
train_ds = BHSDSliceDataset(ROOT, SPLITS['train'], 'train', augment=True, seed=SEED)
val_ds = BHSDSliceDataset(ROOT, SPLITS['val'], 'val', augment=False, seed=SEED)
train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
print('Train slices:', len(train_ds), '| Val slices:', len(val_ds))
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Entrainement --------------------------------------------------------
model = UNet2D().to(DEVICE)
criterion = DiceBCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=LR)
best_dice = 0.0; history = {'train_loss': [], 'val_dice': [], 'val_loss': []}
for epoch in range(EPOCHS):
    model.train(); train_loss = 0
    for imgs, masks in tqdm(train_loader, desc=f'Epoch {epoch+1}/{EPOCHS} [train]', leave=False):
        imgs, masks = imgs.to(DEVICE), masks.to(DEVICE)
        optimizer.zero_grad(); logits = model(imgs); loss = criterion(logits, masks)
        loss.backward(); optimizer.step(); train_loss += loss.item() * imgs.size(0)
    train_loss /= len(train_ds)
    model.eval(); val_loss = 0; val_dice = 0
    with torch.no_grad():
        for imgs, masks in tqdm(val_loader, desc=f'Epoch {epoch+1}/{EPOCHS} [val]', leave=False):
            imgs, masks = imgs.to(DEVICE), masks.to(DEVICE)
            logits = model(imgs); loss = criterion(logits, masks)
            val_loss += loss.item() * imgs.size(0)
            probs = torch.sigmoid(logits); preds = (probs > THRESHOLD).float()
            inter = (preds * masks).sum(); union = preds.sum() + masks.sum()
            val_dice += float((2 * inter + 1e-6) / (union + 1e-6))
    val_loss /= len(val_ds); val_dice /= len(val_loader)
    history['train_loss'].append(train_loss); history['val_dice'].append(val_dice); history['val_loss'].append(val_loss)
    print(f'Epoch {epoch+1:3d} | train={train_loss:.4f} | val={val_loss:.4f} | dice={val_dice:.4f}')
    if val_dice > best_dice:
        best_dice = val_dice; torch.save(model.state_dict(), OUT / 'model_socle.pt')
        print(f'  >>> Meilleur modele (val_dice={best_dice:.4f})')
print(f'\nTermine. Meilleur val_dice : {best_dice:.4f}')
print(f'Modele : {OUT / "model_socle.pt"}')
```
