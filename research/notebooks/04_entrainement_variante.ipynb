```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Dataset avec samplage Lot G -----------------------------------------
class BHSDSliceDatasetLotG(Dataset):
    def __init__(self, root, case_ids, petites_ps, toutes_ps, split, augment=False, seed=36):
        self.root = Path(root); self.case_ids = case_ids; self.petites_ps = petites_ps
        self.toutes_ps = toutes_ps; self.split = split; self.augment = augment; self.rng = random.Random(seed)
        self.slices = []
        for cid in case_ids:
            gt = nib.load(self.root / 'ground truths' / f'{cid}.nii.gz')
            mask = np.asarray(gt.dataobj).astype(np.int16)
            petites = set(petites_ps.get(cid, []))
            for z in range(mask.shape[2]):
                self.slices.append((cid, z, z in petites))
        self.idx_petites = [i for i, (_, _, ep) in enumerate(self.slices) if ep]
        self.idx_autres = [i for i, (_, _, ep) in enumerate(self.slices) if not ep]
        print(f'Total tranches: {len(self.slices)} | Petites: {len(self.idx_petites)} | Autres: {len(self.idx_autres)}')
    def __len__(self): return len(self.slices)
    def __getitem__(self, idx):
        cid, z, est_petite = self.slices[idx]
        img = nib.load(self.root / 'images' / f'{cid}.nii.gz')
        gt = nib.load(self.root / 'ground truths' / f'{cid}.nii.gz')
        image = np.asarray(img.dataobj, dtype=np.float32)
        mask = np.asarray(gt.dataobj).astype(np.int16)
        img_slice = np.clip(image[:, :, z], CT_MIN_HU, CT_MAX_HU)
        img_slice = (img_slice - 40.0) / 80.0  # (valeur - 40) / 80, comme dans l'article BHSD
        mask_slice = mask[:, :, z] > 0
        if self.augment and self.rng.random() < 0.5:
            img_slice = np.fliplr(img_slice); mask_slice = np.fliplr(mask_slice)
        if self.augment:
            from scipy import ndimage
            angle = self.rng.uniform(-10, 10)
            img_slice = ndimage.rotate(img_slice, angle, reshape=False, order=1, mode='nearest')
            mask_slice = ndimage.rotate(mask_slice.astype(np.float32), angle, reshape=False, order=0, mode='nearest') > 0.5
        return (torch.from_numpy(img_slice[None]).float(), torch.from_numpy(mask_slice[None]).float())
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Sampler Lot G : >= 4 tranches petites par batch ---------------------
class LotGBatchSampler:
    def __init__(self, dataset, batch_size=8, shuffle=True, seed=36, n_petites=4):
        self.dataset = dataset
        self.batch_size = batch_size
        self.shuffle = shuffle
        self.rng = random.Random(seed)
        self.n_petites = n_petites
        self.batches = []
        self._rebuild()

    def _rebuild(self):
        """Lots de 8 coupes dont au moins n_petites contiennent une hemorragie < 5 mL.

        Le nombre de lots par epoque est celui du socle (toutes les coupes / 8)
        pour que la comparaison socle/variante ne soit pas biaisee par le temps
        d'entrainement. Les coupes petites reviennent ~1,8 fois par epoque, les
        autres places sont tirees au hasard dans tout le jeu d'entrainement.
        """
        indices = list(range(len(self.dataset)))
        petites = [i for i in indices if self.dataset.slices[i][2]]
        if not petites:
            raise RuntimeError('Aucune coupe < 5 mL : sampler Lot G impossible.')
        if self.shuffle:
            self.rng.shuffle(petites)
        n_p = self.n_petites
        n_a = self.batch_size - n_p
        n_batches = (len(indices) + self.batch_size - 1) // self.batch_size  # comme le socle

        batches = []
        for b in range(n_batches):
            # n_p coupes petites garanties, en cyclant sur le pool melange
            choix_p = [petites[(b * n_p + k) % len(petites)] for k in range(n_p)]
            # n_a coupes tirees au hasard dans tout le dataset (peuvent etre petites)
            choix_a = self.rng.sample(indices, min(n_a, len(indices))) if n_a > 0 else []
            batch = choix_p + choix_a
            if self.shuffle:
                self.rng.shuffle(batch)
            batches.append(batch)
        self.batches = batches

    def __iter__(self):
        if self.shuffle:
            self._rebuild()
        return iter(self.batches)

    def __len__(self):
        return len(self.batches)

```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- U-Net 2D ------------------------------------------------------------
class DoubleConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.net = nn.Sequential(nn.Conv2d(in_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True), nn.Conv2d(out_ch, out_ch, 3, padding=1), nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True))
    def forward(self, x): return self.net(x)

class UNet2D(nn.Module):
    def __init__(self, in_ch=1, out_ch=1, features=(32, 64, 128, 256)):
        super().__init__()
        self.downs = nn.ModuleList(); self.ups = nn.ModuleList(); self.pool = nn.MaxPool2d(2)
        ch = in_ch
        for f in features: self.downs.append(DoubleConv(ch, f)); ch = f
        self.bottleneck = DoubleConv(ch, ch * 2); ch = ch * 2
        for f in reversed(features):
            self.ups.append(nn.ConvTranspose2d(ch, f, 2, stride=2))
            self.ups.append(DoubleConv(f * 2, f)); ch = f
        self.final = nn.Conv2d(features[0], out_ch, 1)
    def forward(self, x):
        skip = []
        for down in self.downs: x = down(x); skip.append(x); x = self.pool(x)
        x = self.bottleneck(x); skip = skip[::-1]
        for i in range(0, len(self.ups), 2):
            x = self.ups[i](x); s = skip[i // 2]
            if x.shape[-2:] != s.shape[-2:]: x = F.interpolate(x, size=s.shape[-2:], mode='bilinear', align_corners=False)
            x = torch.cat([x, s], dim=1); x = self.ups[i + 1](x)
        return self.final(x)
```
