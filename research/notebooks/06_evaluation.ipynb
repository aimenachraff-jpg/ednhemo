```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Configuration -------------------------------------------------------
import os, json, shutil, subprocess
from pathlib import Path
import numpy as np
import pandas as pd
import nibabel as nib
import matplotlib.pyplot as plt

EQUIPE = '36'

if Path('/kaggle/working').is_dir(): BASE = Path('/kaggle/working')
else: BASE = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
ROOT = utils.find_bhsd_root()
SPLITS = utils.load_splits(root=ROOT)

REF_DIR = ROOT / 'ground truths'
# evaluation_commune.py exige une prediction pour CHAQUE fichier du dossier --ref.
# Les predictions ne couvrent que les 96 scanners de test : on isole ces references.
import gzip
REF_TEST = BASE / 'ref_test'
REF_TEST.mkdir(parents=True, exist_ok=True)
for _cid in SPLITS['test']:
    _dst = REF_TEST / f'{_cid}.nii.gz'
    if _dst.exists():
        continue
    # Kaggle decompresse les .nii.gz : la source peut etre .nii.gz ou .nii.
    _src = None
    for _ext in ('.nii.gz', '.nii'):
        _p = REF_DIR / f'{_cid}{_ext}'
        if _p.is_file():
            _src = _p
            break
    if _src is None:
        raise FileNotFoundError(f'Reference {_cid} introuvable dans {REF_DIR}')
    if _src.name.endswith('.nii.gz'):
        shutil.copy2(_src, _dst)
    else:
        # evaluation_commune.py tourne dans un sous-processus (nib.load standard) :
        # le .nii.gz doit etre un vrai fichier gzip.
        with open(_src, 'rb') as _f:
            _donnees = _f.read()
        with gzip.open(_dst, 'wb', compresslevel=1) as _g:
            _g.write(_donnees)
REF_DIR = REF_TEST
print('References test :', len(list(REF_TEST.glob('*.nii.gz'))))

# Kaggle auto-extrait les zips et decompresse les .nii.gz en .nii : le dataset
# monte contient <config>/ID_xxx.nii. evaluation_commune.py exige le nom exact
# ID_xxx.nii.gz : on regzipe les .nii sous le nom attendu dans /kaggle/working.
PRED_INPUTS = {
    'socle': Path('/kaggle/input/m2-predictions-socle-equipe-36'),
    'variante': Path('/kaggle/input/m2-predictions-variante-equipe-36'),
}
PRED_BASE = BASE / 'predictions'
PRED_DIRS = {'socle': PRED_BASE / 'socle', 'variante': PRED_BASE / 'variante'}

def _regzipper(source, dest):
    import gzip
    with open(source, 'rb') as f:
        donnees = f.read()
    with gzip.open(dest, 'wb', compresslevel=1) as g:
        g.write(donnees)

def trouver_predictions(nom):
    dest = PRED_DIRS[nom]
    dest.mkdir(parents=True, exist_ok=True)
    if len(list(dest.glob('*.nii.gz'))) > 0:
        return dest
    racine = PRED_INPUTS[nom]
    # Cas 1 : une archive zip a extraire (si Kaggle ne l'a pas faite).
    for a in sorted(racine.rglob('*.zip')):
        print('Extraction de', a.name)
        import zipfile
        with zipfile.ZipFile(a) as z:
            z.extractall(PRED_BASE)
        if len(list(dest.glob('*.nii.gz'))) > 0:
            return dest
    # Cas 2 : masques .nii.gz nus quelque part dans le dataset.
    if len(list(racine.rglob('*.nii.gz'))) > 0:
        return racine
    # Cas 3 : masques .nii (decompresses par Kaggle) -> regzip sous .nii.gz.
    nus = sorted(racine.rglob('*.nii'))
    if nus:
        for f in nus:
            cible = dest / (f.stem + '.nii.gz')
            if not cible.exists():
                _regzipper(f, cible)
        print(f'{nom} : {len(nus)} masques .nii regzipes en .nii.gz')
    return dest

PRED_SOCLE = trouver_predictions('socle')
PRED_VARIANTE = trouver_predictions('variante')
print('Pred socle    :', PRED_SOCLE, '->', len(list(PRED_SOCLE.glob('*.nii.gz'))), 'masques')
print('Pred variante :', PRED_VARIANTE, '->', len(list(PRED_VARIANTE.glob('*.nii.gz'))), 'masques')
OUT = BASE / 'evaluation'; OUT.mkdir(parents=True, exist_ok=True)
print('Reference :', REF_DIR)
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Comparer ------------------------------------------------------------
res_s = pd.read_csv(OUT / 'resultats_socle_resume.csv')
res_v = pd.read_csv(OUT / 'resultats_variante_resume.csv')
comp = res_s.merge(res_v, on=['critere', 'groupe'], suffixes=('_socle', '_variante'))

dice_s = res_s[res_s['critere'] == 'dice_moyen']['valeur'].values[0]
dice_v = res_v[res_v['critere'] == 'dice_moyen']['valeur'].values[0]
fa_s = res_s[res_s['critere'] == 'fausses_alarmes_par_scanner']['valeur'].values[0]
fa_v = res_v[res_v['critere'] == 'fausses_alarmes_par_scanner']['valeur'].values[0]

print('\n=== COMPARAISON ===')
print(f'Dice moyen      : socle={dice_s:.4f} variante={dice_v:.4f} diff={dice_v-dice_s:+.4f}')
print(f'Fausses alarmes : socle={fa_s:.4f} variante={fa_v:.4f} diff={fa_v-fa_s:+.4f}')

cols = ['critere', 'groupe', 'sens_lesions_socle', 'sens_lesions_variante',
        'sens_par_scanner_socle', 'sens_par_scanner_variante']
cols = [c for c in cols if c in comp.columns]
print(comp[cols].to_string(index=False))
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Figures : sensibilite par taille et par sous-type -------------------
# Les noms de classes sont lus dans les CSV eux-memes (robuste a l'encodage).
fig, axes = plt.subplots(1, 2, figsize=(12.5, 4.6))
width = 0.38

ts = res_s[res_s['critere'] == 'taille']
tv = res_v[res_v['critere'] == 'taille']
classes = ts['groupe'].dropna().tolist()
x = np.arange(len(classes))
ss = [ts[ts['groupe'] == c]['sens_lesions'].values[0] for c in classes]
sv = [tv[tv['groupe'] == c]['sens_lesions'].values[0] for c in classes]
axes[0].bar(x - width / 2, ss, width, label='Socle', color='#4C72B0')
axes[0].bar(x + width / 2, sv, width, label='Variante', color='#DD8452')
axes[0].set_xticks(x)
axes[0].set_xticklabels(classes, rotation=30, ha='right')
axes[0].set_ylabel('Sensibilite')
axes[0].set_title('Par classe de taille')
axes[0].set_ylim(0, 1)
axes[0].legend()
axes[0].grid(alpha=0.3, axis='y')

esp = res_s[res_s['critere'] == 'sous_type']
esv = res_v[res_v['critere'] == 'sous_type']
stypes = esp['groupe'].dropna().tolist()
x2 = np.arange(len(stypes))
s2 = [esp[esp['groupe'] == c]['sens_lesions'].values[0] for c in stypes]
v2 = [esv[esv['groupe'] == c]['sens_lesions'].values[0] for c in stypes]
axes[1].bar(x2 - width / 2, s2, width, label='Socle', color='#4C72B0')
axes[1].bar(x2 + width / 2, v2, width, label='Variante', color='#DD8452')
axes[1].set_xticks(x2)
axes[1].set_xticklabels(stypes, rotation=30, ha='right')
axes[1].set_title('Par sous-type')
axes[1].set_ylim(0, 1)
axes[1].legend()
axes[1].grid(alpha=0.3, axis='y')

fig.suptitle('Comparaison Socle vs Variante (sensibilite par lesion)')
fig.tight_layout()
fig.savefig(OUT / 'comparaison.png', dpi=150)
plt.show()
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Exemples de lesions ratees ------------------------------------------
les_s = pd.read_csv(OUT / 'resultats_socle_lesions.csv')
les_v = pd.read_csv(OUT / 'resultats_variante_lesions.csv')
missed_s = les_s[les_s['detectee'] == 0]
missed_v = les_v[les_v['detectee'] == 0]
print(f'Ratees par socle : {len(missed_s)}')
print(f'Ratees par variante : {len(missed_v)}')

det_s = set(zip(les_s[les_s['detectee'] == 1]['scanner'], les_s[les_s['detectee'] == 1]['lesion']))
det_v = set(zip(les_v[les_v['detectee'] == 1]['scanner'], les_v[les_v['detectee'] == 1]['lesion']))
recup = det_v - det_s
perdues = det_s - det_v
print(f'Recuperees par variante : {len(recup)}')
print(f'Perdues par variante : {len(perdues)}')
```

```python
# --- Charger utils comme module (robuste sous papermill) -----------------

import sys, types

_utils_mod = types.ModuleType('utils')

_utils_mod.__file__ = '/kaggle/working/utils.py'

exec(open('/kaggle/working/utils.py').read(), _utils_mod.__dict__)

sys.modules['utils'] = _utils_mod

utils = _utils_mod

# --- Sauver le resume ----------------------------------------------------
comparison = {'equipe': EQUIPE, 'dice_socle': float(dice_s), 'dice_variante': float(dice_v),
    'dice_diff': float(dice_v - dice_s), 'fa_socle': float(fa_s), 'fa_variante': float(fa_v),
    'fa_diff': float(fa_v - fa_s), 'ratees_socle': int(len(missed_s)),
    'ratees_variante': int(len(missed_v)), 'recuperees': int(len(recup)), 'perdues': int(len(perdues))}

import json as _json
with open(OUT / 'comparaison.json', 'w', encoding='utf-8') as f:
    _json.dump(comparison, f, indent=2, ensure_ascii=False)

print('\n=== RESUME ===')
for k, v in comparison.items(): print(f'  {k}: {v}')
print(f'\nResultats dans : {OUT}')
```
